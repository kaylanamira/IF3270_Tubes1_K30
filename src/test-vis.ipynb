```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from miaw import FFNN
from visualizer import Visualizer

# Inputs for XOR
X = np.array([
    [0, 0],  # class 0
    [0, 1],  # class 1
    [1, 0],  # class 1
    [1, 1]   # class 0
])

# Labels (0 XOR 1 = 1, 1 XOR 1 = 0, etc.)
y = np.array([0, 1, 1, 0])

# One-hot encode the labels for classification
encoder = OneHotEncoder(sparse=False)
y_encoded = encoder.fit_transform(y.reshape(-1, 1))

# You can skip train-test split if you want to visualize the whole thing
X_train, y_train = X, y_encoded
```

```python
# Step 1: Build and train the FFNN
model = FFNN(
    layer_sizes=[2, 4, 2],                   # Input=2, Hidden=4, Output=2
    activations=['tanh', 'softmax'],         # Tanh for hidden, softmax for output
    loss_function='cce',
    weight_init='xavier',
    use_rmsnorm=False                        # Keep it off for clean viz
)

model.fit(X_train, y_train, epochs=50, lr=0.1, batch_size=1, verbose=0)

# Step 2: Visualize using built-in Visualizer class
visualizer = Visualizer(model)
visualizer.plot_neural_network()
```

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.preprocessing import OneHotEncoder

# === 1. Generate clean 2D dataset ===
X, y = make_classification(
    n_samples=200,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_classes=4,
    n_clusters_per_class=1,
    class_sep=4.0,
    random_state=1
)

# === 2. One-hot encode labels ===
encoder = OneHotEncoder(sparse=False)
y_onehot = encoder.fit_transform(y.reshape(-1, 1))
```

```python
model = FFNN(
    layer_sizes=[2, 3, 4],               # 2 input → 6 hidden → 4 output
    activations=["tanh", "softmax"],     # hidden: tanh, output: softmax
    loss_function="cce",                 # multi-class classification
    weight_init="xavier",
    use_rmsnorm=False
)

model.fit(X, y_onehot, epochs=100, lr=0.1, batch_size=16, verbose=1)
```

```python
visualizer = Visualizer(model)
visualizer.plot_neural_network()
```
