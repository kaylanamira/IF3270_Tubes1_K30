```python
import random
from value import Value 
from miawlentera import FFNN  
from sklearn.metrics import accuracy_score
```

```python

def generate_data(num_samples, num_features):
    X = [[Value(random.uniform(-1, 1)) for _ in range(num_features)] for _ in range(num_samples)]
    y = [random.choice([0, 1]) for _ in range(num_samples)]
    return X, y


num_features = 5 
num_train_samples = 500  
num_test_samples = 100 


X_train, y_train = generate_data(num_train_samples, num_features)
X_test, y_test = generate_data(num_test_samples, num_features)

layer_sizes = [num_features, 10,5, 1] 
activations = ['relu','relu' ,'sigmoid']
```

```python
miaurmsmodel = FFNN(layer_sizes, activations,use_rmsnorm=False)
history_l1 = miaurmsmodel.fit(X_train, y_train, epochs=10, batch_size=50, lr=0.05, verbose=0)


miaurmsmodel.save('./models/test-l1-new.json')

load_l1 = FFNN.load('./models/test-l1-new.json')
load_l1pred = load_l1.predict(X_test)
load_l1acc = accuracy_score(y_test, load_l1pred)

print(f"Loaded Miawlentera with L1 Normalization Accuracy: {load_l1acc * 100:.2f}%")
```

```python
history_l2 = miaurmsmodel.fit(X_train,y_train,epochs=10,batch_size=50,lr=0.05,verbose=0, regularization_type="l2", lambda_reg=0.01)
miaurmsmodel.save('./models/test-l2-new.json')

load_l2 = FFNN.load('./models/test-l2-new.json')

load_l2pred = load_l2.predict(X_test)
load_l2acc = accuracy_score(y_test,load_l2pred)
print(f"Loaded Miawlentera with l2 Normalization Accuracy: {load_l2acc * 100:.2f}%")
```

```python
miawlenteraModel = FFNN(layer_sizes, activations,use_rmsnorm=False)

history = miawlenteraModel.fit(X_train, y_train, epochs=10, batch_size=50, lr=0.05, verbose=0)

miaw_pred = miawlenteraModel.predict(X_test)

miaw_acc = accuracy_score(y_test, miaw_pred)


print(f"Miawlentera Accuracy: {miaw_acc * 100:.2f}%")
```

```python
miaurmsmodel = FFNN(layer_sizes, activations,use_rmsnorm=True)

history = miaurmsmodel.fit(X_train,y_train,epochs=10,batch_size=50,lr=0.05,verbose=0)
miaurmsmodel.save('./models/test-rms-new.json')

loadrms = FFNN.load('./models/test-rms-new.json')

loadrmspred = loadrms.predict(X_test)
loadrmsacc = accuracy_score(y_test,loadrmspred)
print(f"Loaded Miawlentera with RMS Normalization Accuracy: {loadrmsacc * 100:.2f}%")
```

```python
import numpy as np

# Function to convert Value objects to raw numbers
def extract_values(X):
    return np.array([[val.data for val in row] for row in X])

X_train_numeric = extract_values(X_train)
X_test_numeric = extract_values(X_test)
```

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(
    hidden_layer_sizes=(num_features,10,5,1),
    learning_rate='constant',
    activation='relu',
    solver='sgd',
    batch_size=50,
    max_iter=10
)

mlp.fit(X_train_numeric, y_train)
y_pred = mlp.predict(X_test_numeric)
acc = accuracy_score(y_test, y_pred)

print(f"MLPClassifier Accuracy: {acc * 100:.2f}%")
```
